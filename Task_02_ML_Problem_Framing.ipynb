```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y = df["churned"]

majority_class = y.mode()[0]
baseline_predictions = np.full(len(y), majority_class)

print("Majority class:", majority_class)
print("Baseline Accuracy:", accuracy_score(y, baseline_predictions))
print("Baseline Precision:", precision_score(y, baseline_predictions, zero_division=0))
print("Baseline Recall:", recall_score(y, baseline_predictions, zero_division=0))
print("Baseline F1:", f1_score(y, baseline_predictions, zero_division=0))
```
